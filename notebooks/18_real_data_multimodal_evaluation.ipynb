```python
# ================================================================
# REAL DATA MULTIMODAL EVALUATION
# AI-Based Instagram Engagement Prediction and Content Optimization System
# ================================================================

import json
import warnings
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

warnings.filterwarnings("ignore")


# ================================================================
# 1. PROJECT CONFIGURATION
# ================================================================

PROJECT_ROOT = Path(
    r"d:\newwwwwwww\AiBasedInstagramPrediction"
)

REAL_FEATURE_DIR = (
    PROJECT_ROOT
    / "datasets"
    / "processed"
    / "real_image_features"
)

RESULTS_DIR = (
    PROJECT_ROOT
    / "results"
    / "real_multimodal_evaluation"
)

RESULTS_DIR.mkdir(
    parents=True,
    exist_ok=True
)


print("=" * 70)
print("REAL DATA MULTIMODAL EVALUATION")
print("=" * 70)

print(
    "\nProject:",
    "AI-Based Instagram Engagement Prediction and Content Optimization System"
)

print("\nProject root:")
print(PROJECT_ROOT)


# ================================================================
# 2. LOAD REAL IMAGE FEATURES
# ================================================================

print("\n" + "=" * 70)
print("LOADING REAL IMAGE FEATURES")
print("=" * 70)

FEATURE_FILE = (
    REAL_FEATURE_DIR
    / "real_resnet50_features.npy"
)

METADATA_FILE = (
    REAL_FEATURE_DIR
    / "real_resnet50_metadata.csv"
)

if not FEATURE_FILE.exists():

    raise FileNotFoundError(
        f"Missing feature file:\n{FEATURE_FILE}"
    )

if not METADATA_FILE.exists():

    raise FileNotFoundError(
        f"Missing metadata file:\n{METADATA_FILE}"
    )


image_features = np.load(
    FEATURE_FILE,
    mmap_mode="r"
)

metadata = pd.read_csv(
    METADATA_FILE
)

print(
    "\nFeature matrix:",
    image_features.shape
)

print(
    "Metadata:",
    metadata.shape
)

print(
    "\nMetadata columns:"
)

for column in metadata.columns:

    print(
        f" - {column}"
    )


# ================================================================
# 3. CONSISTENCY CHECK
# ================================================================

print("\n" + "=" * 70)
print("FEATURE / METADATA CONSISTENCY")
print("=" * 70)

if len(metadata) != image_features.shape[0]:

    raise ValueError(
        "Feature count and metadata count do not match."
    )

print(
    "Feature records:",
    image_features.shape[0]
)

print(
    "Metadata records:",
    len(metadata)
)

print(
    "\nCONSISTENCY CHECK: PASSED"
)


# ================================================================
# 4. SEARCH FOR REAL ENGAGEMENT TARGETS
# ================================================================

print("\n" + "=" * 70)
print("REAL ENGAGEMENT TARGET DETECTION")
print("=" * 70)


# Possible engagement-related column names.

engagement_keywords = [

    "engagement",
    "engagement_rate",
    "likes",
    "like_count",
    "comments",
    "comment_count",
    "shares",
    "share_count",
    "saves",
    "save_count",
    "reach",
    "impressions",
    "views",
    "view_count",
    "followers",
    "followers_count",
    "interactions",
    "interaction"

]


engagement_columns = []

for column in metadata.columns:

    column_lower = str(
        column
    ).lower()

    if any(
        keyword in column_lower
        for keyword in engagement_keywords
    ):

        engagement_columns.append(
            column
        )


print(
    "\nPotential engagement columns:"
)

if engagement_columns:

    for column in engagement_columns:

        print(
            "FOUND:",
            column
        )

else:

    print(
        "NONE FOUND"
    )


# ================================================================
# 5. TARGET-LIKE COLUMN DETECTION
# ================================================================

print("\n" + "=" * 70)
print("TARGET-LIKE FIELD DETECTION")
print("=" * 70)


target_keywords = [

    "target",
    "label",
    "class",
    "performance",
    "engagement",
    "reach",
    "likes",
    "comments",
    "impressions",
    "views"

]


target_candidates = []

for column in metadata.columns:

    column_lower = str(
        column
    ).lower()

    if any(
        keyword in column_lower
        for keyword in target_keywords
    ):

        target_candidates.append(
            column
        )


print(
    "\nCandidate fields:"
)

if target_candidates:

    for column in target_candidates:

        print(
            " -",
            column
        )

else:

    print(
        "No target-like fields detected."
    )


# ================================================================
# 6. METADATA OVERVIEW
# ================================================================

print("\n" + "=" * 70)
print("REAL METADATA OVERVIEW")
print("=" * 70)

print(
    "\nNumber of records:",
    len(metadata)
)

print(
    "Number of metadata fields:",
    len(metadata.columns)
)

print(
    "\nData types:"
)

print(
    metadata.dtypes
)


# ================================================================
# 7. MISSING DATA ANALYSIS
# ================================================================

print("\n" + "=" * 70)
print("REAL METADATA MISSING-VALUE ANALYSIS")
print("=" * 70)

missing_df = pd.DataFrame({

    "column":
        metadata.columns,

    "missing_count":
        metadata.isna().sum().values,

    "missing_percentage":
        (
            metadata.isna().mean()
            * 100
        ).values

})

missing_df = missing_df.sort_values(
    "missing_percentage",
    ascending=False
)

print(
    missing_df
)


missing_df.to_csv(
    RESULTS_DIR
    / "real_metadata_missing_values.csv",
    index=False
)


# ================================================================
# 8. DATASET SOURCE ANALYSIS
# ================================================================

print("\n" + "=" * 70)
print("REAL DATASET SOURCE")
print("=" * 70)

if "dataset_source" in metadata.columns:

    source_counts = (
        metadata[
            "dataset_source"
        ]
        .value_counts()
    )

    print(
        source_counts
    )

    source_counts.to_csv(
        RESULTS_DIR
        / "real_dataset_sources.csv"
    )

else:

    print(
        "dataset_source not available."
    )


# ================================================================
# 9. IMAGE FEATURE SUMMARY
# ================================================================

print("\n" + "=" * 70)
print("RESNET-50 FEATURE SUMMARY")
print("=" * 70)

feature_dimension = image_features.shape[1]

print(
    "\nFeature dimension:",
    feature_dimension
)

print(
    "Number of images:",
    image_features.shape[0]
)

# Calculate statistics on a representative sample.

rng = np.random.default_rng(
    42
)

sample_size = min(
    5000,
    len(metadata)
)

sample_indices = rng.choice(
    len(metadata),
    size=sample_size,
    replace=False
)

feature_sample = np.asarray(
    image_features[
        sample_indices
    ],
    dtype=np.float32
)

print(
    "\nAnalysis sample:",
    sample_size
)

print(
    "Mean feature value:",
    float(
        feature_sample.mean()
    )
)

print(
    "Feature standard deviation:",
    float(
        feature_sample.std()
    )

)


# ================================================================
# 10. FEATURE MAGNITUDE ANALYSIS
# ================================================================

print("\n" + "=" * 70)
print("VISUAL FEATURE MAGNITUDE ANALYSIS")
print("=" * 70)

feature_norms = np.linalg.norm(
    feature_sample,
    axis=1
)

norm_statistics = {

    "mean":
        float(
            feature_norms.mean()
        ),

    "std":
        float(
            feature_norms.std()
        ),

    "minimum":
        float(
            feature_norms.min()
        ),

    "median":
        float(
            np.median(
                feature_norms
            )
        ),

    "maximum":
        float(
            feature_norms.max()
        )

}

for key, value in norm_statistics.items():

    print(
        f"{key.capitalize():10s}: {value:.6f}"
    )


# ================================================================
# 11. FEATURE NORM DISTRIBUTION
# ================================================================

plt.figure(
    figsize=(10, 6)
)

plt.hist(
    feature_norms,
    bins=50,
    alpha=0.8
)

plt.xlabel(
    "ResNet-50 Feature Vector Norm"
)

plt.ylabel(
    "Number of Images"
)

plt.title(
    "Distribution of Real Instagram Image Feature Magnitudes"
)

plt.tight_layout()

norm_plot = (
    RESULTS_DIR
    / "real_resnet50_feature_norm_distribution.png"
)

plt.savefig(
    norm_plot,
    dpi=200,
    bbox_inches="tight"
)

plt.show()

print(
    "\nPlot saved:",
    norm_plot
)


# ================================================================
# 12. CAPTION AVAILABILITY
# ================================================================

print("\n" + "=" * 70)
print("CAPTION AVAILABILITY")
print("=" * 70)

if "caption" in metadata.columns:

    captions = (
        metadata[
            "caption"
        ]
        .fillna("")
        .astype(str)
        .str.strip()
    )

    non_empty = (
        captions.str.len() > 0
    )

    print(
        "Non-empty captions:",
        int(non_empty.sum())
    )

    print(
        "Empty captions:",
        int((~non_empty).sum())
    )

    print(
        "Caption availability:",
        f"{non_empty.mean() * 100:.2f}%"
    )

else:

    print(
        "Caption field not available."
    )


# ================================================================
# 13. REAL DATA MULTIMODAL READINESS CHECK
# ================================================================

print("\n" + "=" * 70)
print("REAL DATA MULTIMODAL READINESS")
print("=" * 70)


has_image_features = (
    image_features.shape[1] == 2048
)

has_metadata = (
    len(metadata) > 0
)

has_caption = (
    "caption"
    in metadata.columns
)

has_engagement_target = (
    len(engagement_columns) > 0
)


print(
    "\nImage features available:",
    has_image_features
)

print(
    "Metadata available:",
    has_metadata
)

print(
    "Caption available:",
    has_caption
)

print(
    "Engagement target available:",
    has_engagement_target
)


# ================================================================
# 14. IMPORTANT SCIENTIFIC VALIDATION
# ================================================================

print("\n" + "=" * 70)
print("SCIENTIFIC VALIDATION")
print("=" * 70)


if not has_engagement_target:

    print(
        """
REAL ENGAGEMENT LABELS WERE NOT FOUND
IN THE CURRENT REAL IMAGE METADATA.

Therefore:

1. Real image feature analysis can be performed.
2. Real image clustering can be performed.
3. Real caption analysis can be performed.
4. Real multimodal feature construction can be performed.
5. REAL MODEL ACCURACY CANNOT YET BE CALCULATED.

No synthetic engagement labels will be assigned
to the real images because doing so would produce
an invalid real-data evaluation.
"""
    )

else:

    print(
        """
REAL ENGAGEMENT-RELATED FIELDS WERE FOUND.

The next stage can use these fields to construct
a legitimate real-data target and evaluate the
multimodal model.
"""
    )


# ================================================================
# 15. SAVE READINESS REPORT
# ================================================================

readiness_report = {

    "real_images":
        int(len(metadata)),

    "resnet50_dimension":
        int(feature_dimension),

    "image_features_available":
        bool(has_image_features),

    "metadata_available":
        bool(has_metadata),

    "caption_available":
        bool(has_caption),

    "engagement_target_available":
        bool(has_engagement_target),

    "engagement_columns":
        [
            str(x)
            for x in engagement_columns
        ],

    "real_model_accuracy_available":
        bool(has_engagement_target),

    "synthetic_labels_used":
        False

}


with open(
    RESULTS_DIR
    / "real_multimodal_readiness.json",
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        readiness_report,
        f,
        indent=4
    )


# ================================================================
# 16. FINAL OUTPUT
# ================================================================

print("\n" + "=" * 70)
print("REAL DATA MULTIMODAL EVALUATION STAGE COMPLETED")
print("=" * 70)

print(
    "\nReal images:",
    len(metadata)
)

print(
    "ResNet-50 features:",
    feature_dimension
)

print(
    "Caption available:",
    has_caption
)

print(
    "Real engagement target:",
    has_engagement_target
)

print(
    "\nResults directory:"
)

print(
    RESULTS_DIR
)

print("\n" + "=" * 70)

if has_engagement_target:

    print(
        "READY FOR REAL-DATA MODEL EVALUATION"
    )

else:

    print(
        "REAL IMAGE FEATURES READY"
    )

    print(
        "NEXT REQUIREMENT: REAL ENGAGEMENT/TARGET DATA"
    )

print("=" * 70)
```
