```python
# ============================================================
# 26_FINAL_EDA.ipynb
# AI-BASED INSTAGRAM ENGAGEMENT PREDICTION SYSTEM
# ============================================================

import os
import json
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from IPython.display import display

warnings.filterwarnings("ignore")

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

print("=" * 70)
print("FINAL DATASET - EXPLORATORY DATA ANALYSIS")
print("=" * 70)
```

```python
# ============================================================
# LOAD DATASET
# ============================================================

DATA_PATH = r"D:\newwwwwwww\AiBasedInstagramPrediction\datasets\Synthetic\synthetic_instagram_engagement_dataset_v2.csv"

if not os.path.exists(DATA_PATH):
    raise FileNotFoundError(f"Dataset not found: {DATA_PATH}")

df = pd.read_csv(DATA_PATH)

print("Dataset loaded successfully.")
print(f"Rows    : {df.shape[0]:,}")
print(f"Columns : {df.shape[1]:,}")

display(df.head())
```

```python
# ============================================================
# DATASET STRUCTURE
# ============================================================

print("=" * 70)
print("DATASET STRUCTURE")
print("=" * 70)

print(f"\nDataset shape: {df.shape[0]:,} rows × {df.shape[1]:,} columns")

print("\nData types:")
print("-" * 70)
print(df.dtypes)

print("\nFirst 5 records:")
print("-" * 70)
display(df.head())
```

```python
# ============================================================
# COLUMN SUMMARY
# ============================================================

column_summary = pd.DataFrame({
    "Column": df.columns,
    "Data Type": df.dtypes.astype(str),
    "Non-Null": df.notna().sum(),
    "Missing": df.isna().sum(),
    "Unique Values": df.nunique()
})

display(column_summary)
```

```python
# ============================================================
# MISSING VALUE ANALYSIS
# ============================================================

missing_summary = pd.DataFrame({
    "Missing Count": df.isna().sum(),
    "Missing Percentage": (df.isna().sum() / len(df)) * 100
})

missing_summary = missing_summary[
    missing_summary["Missing Count"] > 0
].sort_values("Missing Count", ascending=False)

print("=" * 70)
print("MISSING VALUE ANALYSIS")
print("=" * 70)

if missing_summary.empty:
    print("\nNo missing values were found in the dataset.")
else:
    display(missing_summary)

    # Visualise missing values
    plt.figure(figsize=(12, 6))

    plt.bar(
        missing_summary.index,
        missing_summary["Missing Percentage"]
    )

    plt.title("Missing Values by Feature")
    plt.xlabel("Feature")
    plt.ylabel("Missing Values (%)")
    plt.xticks(rotation=90)
    plt.tight_layout()
    plt.show()
```

```python
# ============================================================
# DUPLICATE ANALYSIS
# ============================================================

duplicate_count = df.duplicated().sum()
duplicate_percentage = (duplicate_count / len(df)) * 100

print("=" * 70)
print("DUPLICATE ANALYSIS")
print("=" * 70)

print(f"\nTotal duplicate records    : {duplicate_count:,}")
print(f"Duplicate percentage       : {duplicate_percentage:.2f}%")

if duplicate_count == 0:
    print("\nNo duplicate records were identified.")
else:
    print("\nDuplicate records were identified and will be considered during data preparation.")
```

```python
# ============================================================
# TARGET VARIABLE ANALYSIS
# ============================================================

TARGET = "performance_class"

print("=" * 70)
print("TARGET VARIABLE ANALYSIS")
print("=" * 70)

print("\nTarget variable:", TARGET)

target_counts = df[TARGET].value_counts()
target_percentages = df[TARGET].value_counts(normalize=True) * 100

target_summary = pd.DataFrame({
    "Count": target_counts,
    "Percentage": target_percentages.round(2)
})

display(target_summary)
```

```python
# ============================================================
# TARGET DISTRIBUTION
# ============================================================

plt.figure(figsize=(8, 5))

df[TARGET].value_counts().plot(kind="bar")

plt.title("Distribution of Instagram Engagement Performance Classes")
plt.xlabel("Performance Class")
plt.ylabel("Number of Posts")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# TARGET CLASS PROPORTIONS
# ============================================================

plt.figure(figsize=(8, 5))

target_percentages.sort_index().plot(kind="bar")

plt.title("Percentage Distribution of Engagement Performance Classes")
plt.xlabel("Performance Class")
plt.ylabel("Percentage of Posts (%)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# NUMERICAL FEATURE ANALYSIS
# ============================================================

numerical_features = df.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

# Remove target if it is represented numerically
numerical_features = [
    col for col in numerical_features
    if col != TARGET
]

print("=" * 70)
print("NUMERICAL FEATURES")
print("=" * 70)

print(f"\nNumber of numerical features: {len(numerical_features)}")

for feature in numerical_features:
    print("-", feature)
```

```python
# ============================================================
# NUMERICAL FEATURE STATISTICS
# ============================================================

numerical_summary = df[numerical_features].describe().T

numerical_summary["missing"] = df[numerical_features].isna().sum()
numerical_summary["missing_%"] = (
    df[numerical_features].isna().sum() / len(df) * 100
).round(2)

display(numerical_summary)
```

```python
# ============================================================
# ACCOUNT-RELATED FEATURE DISTRIBUTIONS
# ============================================================

account_features = [
    "follower_count",
    "following_count",
    "account_age_days",
    "posting_frequency",
    "average_historical_engagement",
    "audience_growth_rate",
    "account_activity_level",
    "content_consistency"
]

account_features = [
    feature for feature in account_features
    if feature in df.columns
]

for feature in account_features:

    plt.figure(figsize=(8, 5))

    plt.hist(
        df[feature].dropna(),
        bins=30
    )

    plt.title(f"Distribution of {feature.replace('_', ' ').title()}")
    plt.xlabel(feature.replace("_", " ").title())
    plt.ylabel("Frequency")
    plt.tight_layout()
    plt.show()
```

```python
# ============================================================
# CAPTION AND HASHTAG FEATURE DISTRIBUTIONS
# ============================================================

text_features = [
    "caption_length",
    "word_count",
    "sentence_count",
    "hashtag_count",
    "unique_hashtag_count",
    "average_hashtag_length",
    "hashtag_character_count",
    "emoji_count",
    "mention_count",
    "question_mark_count",
    "exclamation_count",
    "uppercase_ratio",
    "numeric_token_count",
    "caption_sentiment",
    "caption_subjectivity",
    "caption_readability",
    "keyword_density",
    "caption_complexity",
    "caption_engagement_intent"
]

text_features = [
    feature for feature in text_features
    if feature in df.columns
]

for feature in text_features:

    plt.figure(figsize=(8, 5))

    plt.hist(
        df[feature].dropna(),
        bins=30
    )

    plt.title(f"Distribution of {feature.replace('_', ' ').title()}")
    plt.xlabel(feature.replace("_", " ").title())
    plt.ylabel("Frequency")
    plt.tight_layout()
    plt.show()
```

```python
# ============================================================
# IMAGE FEATURE DISTRIBUTIONS
# ============================================================

image_features = [
    "image_width",
    "image_height",
    "aspect_ratio",
    "brightness",
    "contrast",
    "saturation",
    "sharpness",
    "colorfulness",
    "face_count",
    "text_in_image",
    "visual_complexity",
    "estimated_image_quality"
]

image_features = [
    feature for feature in image_features
    if feature in df.columns
]

for feature in image_features:

    plt.figure(figsize=(8, 5))

    plt.hist(
        df[feature].dropna(),
        bins=30
    )

    plt.title(f"Distribution of {feature.replace('_', ' ').title()}")
    plt.xlabel(feature.replace("_", " ").title())
    plt.ylabel("Frequency")
    plt.tight_layout()
    plt.show()
```

```python
# ============================================================
# CATEGORICAL FEATURE ANALYSIS
# ============================================================

categorical_features = df.select_dtypes(
    include=["object", "bool", "category"]
).columns.tolist()

# Remove text-heavy and identifier columns
exclude_categorical = [
    "post_id",
    "account_id",
    "caption",
    "hashtags",
    "posting_datetime",
    TARGET
]

categorical_features = [
    feature for feature in categorical_features
    if feature not in exclude_categorical
]

print("=" * 70)
print("CATEGORICAL FEATURES")
print("=" * 70)

print(f"\nNumber of categorical features: {len(categorical_features)}")

for feature in categorical_features:
    print("-", feature)
```

```python
# ============================================================
# CATEGORICAL FEATURE DISTRIBUTIONS
# ============================================================

for feature in categorical_features:

    print(f"\n{feature}")
    print("-" * 50)

    display(df[feature].value_counts().to_frame("Count"))

    plt.figure(figsize=(8, 5))

    df[feature].value_counts().plot(kind="bar")

    plt.title(
        f"Distribution of {feature.replace('_', ' ').title()}"
    )
    plt.xlabel(feature.replace("_", " ").title())
    plt.ylabel("Number of Posts")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()
```

```python
# ============================================================
# CATEGORICAL FEATURES VS PERFORMANCE CLASS
# ============================================================

for feature in categorical_features:

    cross_tab = pd.crosstab(
        df[feature],
        df[TARGET],
        normalize="index"
    ) * 100

    print("=" * 70)
    print(f"{feature.upper()} VS PERFORMANCE CLASS")
    print("=" * 70)

    display(cross_tab.round(2))

    cross_tab.plot(
        kind="bar",
        figsize=(9, 5)
    )

    plt.title(
        f"{feature.replace('_', ' ').title()} vs Performance Class"
    )
    plt.xlabel(feature.replace("_", " ").title())
    plt.ylabel("Percentage of Posts (%)")
    plt.xticks(rotation=45, ha="right")
    plt.legend(title="Performance Class")
    plt.tight_layout()
    plt.show()
```

```python
# ============================================================
# TARGET ENCODING FOR CORRELATION ANALYSIS
# ============================================================

performance_mapping = {
    "Low": 0,
    "Medium": 1,
    "High": 2
}

df["performance_class_encoded"] = (
    df[TARGET].map(performance_mapping)
)

print("Target encoding:")
print(performance_mapping)

display(
    df[[TARGET, "performance_class_encoded"]].head()
)
```

```python
# ============================================================
# CORRELATION ANALYSIS
# ============================================================

correlation_features = numerical_features + [
    "performance_class_encoded"
]

correlation_matrix = df[correlation_features].corr()

target_correlations = (
    correlation_matrix["performance_class_encoded"]
    .drop("performance_class_encoded")
    .sort_values(key=abs, ascending=False)
)

print("=" * 70)
print("FEATURE CORRELATION WITH PERFORMANCE CLASS")
print("=" * 70)

display(
    target_correlations.to_frame("Correlation")
)
```

```python
# ============================================================
# TOP FEATURES CORRELATED WITH PERFORMANCE
# ============================================================

top_correlations = target_correlations.head(15).sort_values()

plt.figure(figsize=(10, 7))

top_correlations.plot(kind="barh")

plt.title(
    "Top 15 Features Correlated with Engagement Performance"
)
plt.xlabel("Correlation with Performance Class")
plt.ylabel("Feature")
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# CORRELATION HEATMAP
# ============================================================

top_features = target_correlations.head(15).index.tolist()

heatmap_features = top_features + [
    "performance_class_encoded"
]

plt.figure(figsize=(12, 9))

plt.imshow(
    df[heatmap_features].corr(),
    aspect="auto"
)

plt.colorbar()

plt.xticks(
    range(len(heatmap_features)),
    heatmap_features,
    rotation=90
)

plt.yticks(
    range(len(heatmap_features)),
    heatmap_features
)

plt.title("Correlation Heatmap of Important Features")

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# OUTLIER ANALYSIS USING IQR
# ============================================================

outlier_summary = []

for feature in numerical_features:

    series = df[feature].dropna()

    Q1 = series.quantile(0.25)
    Q3 = series.quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = series[
        (series < lower_bound) |
        (series > upper_bound)
    ]

    outlier_summary.append({
        "Feature": feature,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Outlier Count": len(outliers),
        "Outlier Percentage": (len(outliers) / len(series)) * 100
    })

outlier_summary = pd.DataFrame(outlier_summary)

outlier_summary = outlier_summary.sort_values(
    "Outlier Count",
    ascending=False
)

display(outlier_summary.round(2))
```

```python
# ============================================================
# TOP FEATURES WITH OUTLIERS
# ============================================================

top_outlier_features = (
    outlier_summary
    .head(10)["Feature"]
    .tolist()
)

plt.figure(figsize=(10, 6))

plt.barh(
    top_outlier_features[::-1],
    outlier_summary.head(10)["Outlier Count"][::-1]
)

plt.title("Top 10 Features by Number of Potential Outliers")
plt.xlabel("Number of Potential Outliers")
plt.ylabel("Feature")
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# BOXPLOTS FOR SELECTED FEATURES
# ============================================================

selected_boxplot_features = [
    "follower_count",
    "following_count",
    "average_historical_engagement",
    "caption_length",
    "hashtag_count",
    "brightness",
    "contrast",
    "sharpness",
    "estimated_image_quality"
]

selected_boxplot_features = [
    feature for feature in selected_boxplot_features
    if feature in df.columns
]

for feature in selected_boxplot_features:

    plt.figure(figsize=(8, 4))

    plt.boxplot(
        df[feature].dropna(),
        vert=False
    )

    plt.title(
        f"Box Plot of {feature.replace('_', ' ').title()}"
    )
    plt.xlabel(feature.replace("_", " ").title())
    plt.tight_layout()
    plt.show()
```

```python
# ============================================================
# EDA SUMMARY
# ============================================================

print("=" * 70)
print("EXPLORATORY DATA ANALYSIS SUMMARY")
print("=" * 70)

print(f"\n1. Dataset size:")
print(f"   {df.shape[0]:,} records and {df.shape[1]:,} columns")

print("\n2. Target variable:")
print(f"   {TARGET}")
print(f"   Classes: {df[TARGET].unique().tolist()}")

print("\n3. Missing values:")
print(f"   Features containing missing values: {len(missing_summary)}")

print("\n4. Duplicate records:")
print(f"   {duplicate_count:,} duplicate records")

print("\n5. Numerical features:")
print(f"   {len(numerical_features)} numerical features")

print("\n6. Categorical features:")
print(f"   {len(categorical_features)} categorical features")

print("\n7. Features with the most potential outliers:")
for feature in top_outlier_features:
    print(f"   - {feature}")
```
