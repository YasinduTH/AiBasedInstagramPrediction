```python
# ============================================================
# 14_MULTIMODAL_IMAGE_INTEGRATION.ipynb
# AI-BASED INSTAGRAM ENGAGEMENT PREDICTION SYSTEM
# ============================================================

import json
import time
import warnings
from pathlib import Path

import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import (
    train_test_split,
    StratifiedKFold,
    cross_validate
)

from sklearn.compose import ColumnTransformer

from sklearn.pipeline import Pipeline

from sklearn.impute import SimpleImputer

from sklearn.preprocessing import OneHotEncoder

from sklearn.ensemble import HistGradientBoostingClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

warnings.filterwarnings("ignore")


# ============================================================
# 14.1 NOTEBOOK INITIALIZATION
# ============================================================

print("=" * 75)
print("MULTIMODAL IMAGE FEATURE INTEGRATION")
print("=" * 75)

print("""
AI-Based Instagram Engagement Prediction and
Content Optimization System

Experiment:
Baseline Text + Metadata
        VS
Text + Metadata + Image Features
""")


# ============================================================
# 14.2 PROJECT PATH DETECTION
# ============================================================

print("\n" + "=" * 75)
print("PROJECT STRUCTURE")
print("=" * 75)

PROJECT_ROOT = Path.cwd()

while PROJECT_ROOT != PROJECT_ROOT.parent:

    if (
        (PROJECT_ROOT / "datasets").exists()
        and (PROJECT_ROOT / "models").exists()
    ):
        break

    PROJECT_ROOT = PROJECT_ROOT.parent


DATASET_PATH = (
    PROJECT_ROOT
    / "datasets"
    / "Synthetic"
    / "synthetic_instagram_engagement_dataset_v2.csv"
)

MODELS_DIR = (
    PROJECT_ROOT
    / "models"
)

RESULTS_DIR = (
    PROJECT_ROOT
    / "results"
)

MODELS_DIR.mkdir(
    parents=True,
    exist_ok=True
)

RESULTS_DIR.mkdir(
    parents=True,
    exist_ok=True
)

print("Project root :", PROJECT_ROOT)
print("Dataset      :", DATASET_PATH)
print("Exists       :", DATASET_PATH.exists())

if not DATASET_PATH.exists():

    raise FileNotFoundError(
        f"Dataset not found:\n{DATASET_PATH}"
    )


# ============================================================
# 14.3 LOAD DATA
# ============================================================

print("\n" + "=" * 75)
print("LOADING SYNTHETIC V2")
print("=" * 75)

df = pd.read_csv(
    DATASET_PATH
)

print(
    "\nRows    :",
    len(df)
)

print(
    "Columns :",
    len(df.columns)
)

TARGET = "performance_class"

if TARGET not in df.columns:

    raise ValueError(
        f"Target column '{TARGET}' not found."
    )


print("\nTarget distribution:")

display(
    df[TARGET]
    .value_counts()
    .to_frame("Count")
)


# ============================================================
# 14.4 LEAKAGE / IDENTIFIER EXCLUSION
# ============================================================

print("\n" + "=" * 75)
print("LEAKAGE AND IDENTIFIER AUDIT")
print("=" * 75)


IDENTIFIER_COLUMNS = [
    "post_id",
    "account_id"
]


POST_PUBLICATION_COLUMNS = [
    "likes",
    "comments",
    "shares",
    "saves",
    "reach",
    "impressions",
    "engagement_rate",
    "binary_performance"
]


RAW_TEXT_COLUMNS = [
    "caption",
    "hashtags",
    "posting_datetime"
]


ALL_EXCLUDED = (
    IDENTIFIER_COLUMNS
    + POST_PUBLICATION_COLUMNS
    + RAW_TEXT_COLUMNS
)


excluded_found = [
    col
    for col in ALL_EXCLUDED
    if col in df.columns
]


print("\nColumns excluded from modelling:")

for col in excluded_found:

    print(
        "  -",
        col
    )


# ============================================================
# 14.5 IDENTIFY IMAGE FEATURES
# ============================================================

print("\n" + "=" * 75)
print("IMAGE FEATURE IDENTIFICATION")
print("=" * 75)


# These are the image-derived features available
# in Synthetic V2.

IMAGE_FEATURES = [

    "brightness",
    "contrast",
    "saturation",
    "sharpness",
    "colorfulness",
    "face_count",
    "text_in_image",
    "visual_complexity",
    "estimated_image_quality"
]


image_features_found = [
    col
    for col in IMAGE_FEATURES
    if col in df.columns
]


image_features_missing = [
    col
    for col in IMAGE_FEATURES
    if col not in df.columns
]


print("\nImage features found:")

for col in image_features_found:

    print(
        "  ✓",
        col
    )


print("\nImage features missing:")

if image_features_missing:

    for col in image_features_missing:

        print(
            "  -",
            col
        )

else:

    print(
        "  None"
    )


print(
    "\nTotal image features:",
    len(image_features_found)
)


if len(image_features_found) == 0:

    raise ValueError(
        "No image-derived features were found in Synthetic V2."
    )


# ============================================================
# 14.6 CREATE BASELINE AND MULTIMODAL FEATURE SETS
# ============================================================

print("\n" + "=" * 75)
print("FEATURE GROUP CONSTRUCTION")
print("=" * 75)


# Remove leakage, identifiers, raw text fields and target.

base_features = [
    col
    for col in df.columns
    if col not in excluded_found
    and col != TARGET
]


# Remove image features from baseline.

baseline_features = [
    col
    for col in base_features
    if col not in image_features_found
]


# Multimodal = baseline + image features.

multimodal_features = (
    baseline_features
    + image_features_found
)


print(
    "\nBaseline features:",
    len(baseline_features)
)

print(
    "Image features:",
    len(image_features_found)
)

print(
    "Multimodal features:",
    len(multimodal_features)
)


# Safety check

if TARGET in baseline_features:

    raise ValueError(
        "Target accidentally included in baseline features."
    )


if TARGET in multimodal_features:

    raise ValueError(
        "Target accidentally included in multimodal features."
    )


for prohibited in (
    IDENTIFIER_COLUMNS
    + POST_PUBLICATION_COLUMNS
):

    if prohibited in baseline_features:

        raise ValueError(
            f"Leakage column in baseline: {prohibited}"
        )

    if prohibited in multimodal_features:

        raise ValueError(
            f"Leakage column in multimodal: {prohibited}"
        )


print(
    "\nFEATURE CONSTRUCTION CHECK: PASSED"
)


# ============================================================
# 14.7 IMAGE FEATURE QUALITY AUDIT
# ============================================================

print("\n" + "=" * 75)
print("IMAGE FEATURE QUALITY AUDIT")
print("=" * 75)


image_quality_rows = []

for col in image_features_found:

    image_quality_rows.append({

        "Feature": col,

        "Data_Type":
            str(df[col].dtype),

        "Missing_Count":
            int(df[col].isna().sum()),

        "Missing_Percentage":
            float(
                df[col].isna().mean() * 100
            ),

        "Unique_Values":
            int(
                df[col].nunique(
                    dropna=True
                )
            ),

        "Mean":
            float(
                df[col].mean()
            ),

        "Std":
            float(
                df[col].std()
            ),

        "Minimum":
            float(
                df[col].min()
            ),

        "Maximum":
            float(
                df[col].max()
            )
    })


image_quality_df = pd.DataFrame(
    image_quality_rows
)

display(
    image_quality_df.round(4)
)


image_quality_path = (
    RESULTS_DIR
    / "image_feature_quality_audit.csv"
)

image_quality_df.to_csv(
    image_quality_path,
    index=False
)


# ============================================================
# 14.8 IMAGE FEATURE → TARGET ANALYSIS
# ============================================================

print("\n" + "=" * 75)
print("IMAGE FEATURE CLASS ANALYSIS")
print("=" * 75)


class_means = (
    df.groupby(TARGET)[
        image_features_found
    ]
    .mean()
)


display(
    class_means.round(4)
)


class_means_path = (
    RESULTS_DIR
    / "image_feature_class_means.csv"
)

class_means.to_csv(
    class_means_path
)


# ============================================================
# 14.9 PREPARE TARGET
# ============================================================

y = (
    df[TARGET]
    .astype(str)
)


# ============================================================
# 14.10 REPRODUCIBLE TRAIN / TEST SPLIT
# ============================================================

print("\n" + "=" * 75)
print("TRAIN / TEST SPLIT")
print("=" * 75)


train_indices, test_indices = train_test_split(
    np.arange(len(df)),
    test_size=0.20,
    stratify=y,
    random_state=42
)


print(
    "\nTraining records:",
    len(train_indices)
)

print(
    "Testing records :",
    len(test_indices)
)


# ============================================================
# 14.11 MODEL CONFIGURATION
# ============================================================

MODEL_PARAMETERS = {

    "learning_rate": 0.08,

    "max_iter": 300,

    "max_leaf_nodes": 63,

    "min_samples_leaf": 20,

    "l2_regularization": 0.1,

    "random_state": 42
}


def build_pipeline(
    feature_columns
):

    X = (
        df[
            feature_columns
        ]
        .copy()
    )


    numeric_features = (
        X.select_dtypes(
            include=["number"]
        )
        .columns
        .tolist()
    )


    categorical_features = (
        X.select_dtypes(
            include=[
                "object",
                "category",
                "string"
            ]
        )
        .columns
        .tolist()
    )


    # Boolean → integer

    for col in X.columns:

        if X[col].dtype == bool:

            X[col] = X[col].astype(int)


    # Numeric conversion

    for col in numeric_features:

        X[col] = pd.to_numeric(
            X[col],
            errors="coerce"
        )


    # Categorical conversion

    for col in categorical_features:

        X[col] = X[col].astype("string")


    numeric_pipeline = Pipeline([

        (
            "imputer",
            SimpleImputer(
                strategy="median"
            )
        )
    ])


    categorical_pipeline = Pipeline([

        (
            "imputer",
            SimpleImputer(
                strategy="most_frequent"
            )
        ),

        (
            "encoder",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ])


    preprocessor = ColumnTransformer([

        (
            "numeric",
            numeric_pipeline,
            numeric_features
        ),

        (
            "categorical",
            categorical_pipeline,
            categorical_features
        )
    ])


    model = HistGradientBoostingClassifier(
        **MODEL_PARAMETERS
    )


    pipeline = Pipeline([

        (
            "preprocessor",
            preprocessor
        ),

        (
            "classifier",
            model
        )
    ])


    return X, pipeline


# ============================================================
# 14.12 BASELINE MODEL
# ============================================================

print("\n" + "=" * 75)
print("MODEL A — TEXT + METADATA BASELINE")
print("=" * 75)


X_baseline, baseline_pipeline = build_pipeline(
    baseline_features
)


X_base_train = (
    X_baseline
    .iloc[train_indices]
)

X_base_test = (
    X_baseline
    .iloc[test_indices]
)

y_train = (
    y.iloc[train_indices]
)

y_test = (
    y.iloc[test_indices]
)


print(
    "\nTraining baseline model..."
)

start = time.time()

baseline_pipeline.fit(
    X_base_train,
    y_train
)

baseline_training_time = (
    time.time() - start
)


baseline_predictions = (
    baseline_pipeline
    .predict(X_base_test)
)


baseline_accuracy = accuracy_score(
    y_test,
    baseline_predictions
)

baseline_precision = precision_score(
    y_test,
    baseline_predictions,
    average="weighted",
    zero_division=0
)

baseline_recall = recall_score(
    y_test,
    baseline_predictions,
    average="weighted",
    zero_division=0
)

baseline_f1 = f1_score(
    y_test,
    baseline_predictions,
    average="weighted",
    zero_division=0
)

baseline_macro_f1 = f1_score(
    y_test,
    baseline_predictions,
    average="macro",
    zero_division=0
)


print(
    "\nBASELINE RESULTS"
)

print(
    f"Accuracy           : {baseline_accuracy:.4%}"
)

print(
    f"Weighted Precision : {baseline_precision:.4%}"
)

print(
    f"Weighted Recall    : {baseline_recall:.4%}"
)

print(
    f"Weighted F1        : {baseline_f1:.4%}"
)

print(
    f"Macro F1           : {baseline_macro_f1:.4%}"
)

print(
    f"Training Time      : {baseline_training_time:.2f}s"
)


# ============================================================
# 14.13 MULTIMODAL MODEL
# ============================================================

print("\n" + "=" * 75)
print("MODEL B — TEXT + METADATA + IMAGE FEATURES")
print("=" * 75)


X_multimodal, multimodal_pipeline = build_pipeline(
    multimodal_features
)


X_multi_train = (
    X_multimodal
    .iloc[train_indices]
)

X_multi_test = (
    X_multimodal
    .iloc[test_indices]
)


print(
    "\nTraining multimodal model..."
)

start = time.time()

multimodal_pipeline.fit(
    X_multi_train,
    y_train
)

multimodal_training_time = (
    time.time() - start
)


multimodal_predictions = (
    multimodal_pipeline
    .predict(X_multi_test)
)


multimodal_accuracy = accuracy_score(
    y_test,
    multimodal_predictions
)

multimodal_precision = precision_score(
    y_test,
    multimodal_predictions,
    average="weighted",
    zero_division=0
)

multimodal_recall = recall_score(
    y_test,
    multimodal_predictions,
    average="weighted",
    zero_division=0
)

multimodal_f1 = f1_score(
    y_test,
    multimodal_predictions,
    average="weighted",
    zero_division=0
)

multimodal_macro_f1 = f1_score(
    y_test,
    multimodal_predictions,
    average="macro",
    zero_division=0
)


print(
    "\nMULTIMODAL RESULTS"
)

print(
    f"Accuracy           : {multimodal_accuracy:.4%}"
)

print(
    f"Weighted Precision : {multimodal_precision:.4%}"
)

print(
    f"Weighted Recall    : {multimodal_recall:.4%}"
)

print(
    f"Weighted F1        : {multimodal_f1:.4%}"
)

print(
    f"Macro F1           : {multimodal_macro_f1:.4%}"
)

print(
    f"Training Time      : {multimodal_training_time:.2f}s"
)


# ============================================================
# 14.14 IMAGE CONTRIBUTION
# ============================================================

print("\n" + "=" * 75)
print("IMAGE FEATURE CONTRIBUTION")
print("=" * 75)


accuracy_change = (
    multimodal_accuracy
    - baseline_accuracy
)

f1_change = (
    multimodal_f1
    - baseline_f1
)


accuracy_change_percentage_points = (
    accuracy_change * 100
)

f1_change_percentage_points = (
    f1_change * 100
)


print(
    f"\nBaseline Accuracy   : "
    f"{baseline_accuracy:.4%}"
)

print(
    f"Multimodal Accuracy : "
    f"{multimodal_accuracy:.4%}"
)

print(
    f"Accuracy Change     : "
    f"{accuracy_change_percentage_points:+.2f} percentage points"
)


print(
    f"\nBaseline Weighted F1   : "
    f"{baseline_f1:.4%}"
)

print(
    f"Multimodal Weighted F1 : "
    f"{multimodal_f1:.4%}"
)

print(
    f"F1 Change              : "
    f"{f1_change_percentage_points:+.2f} percentage points"
)


if multimodal_accuracy > baseline_accuracy:

    image_effect = "IMPROVED"

elif multimodal_accuracy < baseline_accuracy:

    image_effect = "DECREASED"

else:

    image_effect = "NO CHANGE"


print(
    "\nImage integration result:",
    image_effect
)


# ============================================================
# 14.15 MODEL COMPARISON TABLE
# ============================================================

comparison = pd.DataFrame({

    "Model": [

        "Text + Metadata Baseline",

        "Text + Metadata + Image Features"
    ],

    "Feature_Count": [

        len(baseline_features),

        len(multimodal_features)
    ],

    "Image_Feature_Count": [

        0,

        len(image_features_found)
    ],

    "Accuracy": [

        baseline_accuracy,

        multimodal_accuracy
    ],

    "Weighted_Precision": [

        baseline_precision,

        multimodal_precision
    ],

    "Weighted_Recall": [

        baseline_recall,

        multimodal_recall
    ],

    "Weighted_F1": [

        baseline_f1,

        multimodal_f1
    ],

    "Macro_F1": [

        baseline_macro_f1,

        multimodal_macro_f1
    ],

    "Training_Time_Seconds": [

        baseline_training_time,

        multimodal_training_time
    ]
})


print("\n" + "=" * 75)
print("MODEL COMPARISON")
print("=" * 75)

display(
    comparison.round(4)
)


comparison_path = (
    RESULTS_DIR
    / "multimodal_model_comparison.csv"
)

comparison.to_csv(
    comparison_path,
    index=False
)


# ============================================================
# 14.16 CLASSIFICATION REPORT — MULTIMODAL
# ============================================================

print("\n" + "=" * 75)
print("MULTIMODAL CLASSIFICATION REPORT")
print("=" * 75)


multimodal_report = classification_report(
    y_test,
    multimodal_predictions,
    zero_division=0
)

print(
    multimodal_report
)


multimodal_report_dict = classification_report(
    y_test,
    multimodal_predictions,
    output_dict=True,
    zero_division=0
)


multimodal_report_df = (
    pd.DataFrame(
        multimodal_report_dict
    )
    .transpose()
)


display(
    multimodal_report_df.round(4)
)


multimodal_report_path = (
    RESULTS_DIR
    / "multimodal_classification_report.csv"
)

multimodal_report_df.to_csv(
    multimodal_report_path
)


# ============================================================
# 14.17 MULTIMODAL CONFUSION MATRIX
# ============================================================

print("\n" + "=" * 75)
print("MULTIMODAL CONFUSION MATRIX")
print("=" * 75)


classes = sorted(
    y.unique()
)


multimodal_cm = confusion_matrix(
    y_test,
    multimodal_predictions,
    labels=classes
)


multimodal_cm_df = pd.DataFrame(

    multimodal_cm,

    index=[
        f"Actual_{c}"
        for c in classes
    ],

    columns=[
        f"Predicted_{c}"
        for c in classes
    ]
)


display(
    multimodal_cm_df
)


multimodal_cm_path = (
    RESULTS_DIR
    / "multimodal_confusion_matrix.csv"
)

multimodal_cm_df.to_csv(
    multimodal_cm_path
)


# ============================================================
# 14.18 CONFUSION MATRIX PLOT
# ============================================================

plt.figure(
    figsize=(8, 6)
)

plt.imshow(
    multimodal_cm
)

plt.title(
    "Multimodal Model Confusion Matrix"
)

plt.xlabel(
    "Predicted Class"
)

plt.ylabel(
    "Actual Class"
)

plt.xticks(
    range(len(classes)),
    classes
)

plt.yticks(
    range(len(classes)),
    classes
)

for i in range(len(classes)):

    for j in range(len(classes)):

        plt.text(
            j,
            i,
            str(multimodal_cm[i, j]),
            ha="center",
            va="center"
        )


plt.tight_layout()


multimodal_cm_image = (
    RESULTS_DIR
    / "multimodal_confusion_matrix.png"
)

plt.savefig(
    multimodal_cm_image,
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 14.19 5-FOLD CROSS-VALIDATION
# ============================================================

print("\n" + "=" * 75)
print("5-FOLD CROSS-VALIDATION")
print("=" * 75)

print("""
Running the same five-fold validation procedure
for both feature configurations.
""")


cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)


# ------------------------------------------------------------
# Baseline CV
# ------------------------------------------------------------

print(
    "\nRunning baseline CV..."
)

cv_start = time.time()

baseline_cv = cross_validate(

    baseline_pipeline,

    X_baseline,

    y,

    cv=cv,

    scoring={
        "accuracy": "accuracy",
        "weighted_f1": "f1_weighted"
    },

    return_train_score=True,

    n_jobs=1
)


baseline_cv_time = (
    time.time()
    - cv_start
)


# ------------------------------------------------------------
# Multimodal CV
# ------------------------------------------------------------

print(
    "Running multimodal CV..."
)

cv_start = time.time()

multimodal_cv = cross_validate(

    multimodal_pipeline,

    X_multimodal,

    y,

    cv=cv,

    scoring={
        "accuracy": "accuracy",
        "weighted_f1": "f1_weighted"
    },

    return_train_score=True,

    n_jobs=1
)


multimodal_cv_time = (
    time.time()
    - cv_start
)


# ============================================================
# 14.20 CV SUMMARY
# ============================================================

baseline_cv_accuracy = (
    baseline_cv[
        "test_accuracy"
    ].mean()
)

baseline_cv_accuracy_std = (
    baseline_cv[
        "test_accuracy"
    ].std()
)

baseline_cv_f1 = (
    baseline_cv[
        "test_weighted_f1"
    ].mean()
)

baseline_cv_f1_std = (
    baseline_cv[
        "test_weighted_f1"
    ].std()
)


multimodal_cv_accuracy = (
    multimodal_cv[
        "test_accuracy"
    ].mean()
)

multimodal_cv_accuracy_std = (
    multimodal_cv[
        "test_accuracy"
    ].std()
)

multimodal_cv_f1 = (
    multimodal_cv[
        "test_weighted_f1"
    ].mean()
)

multimodal_cv_f1_std = (
    multimodal_cv[
        "test_weighted_f1"
    ].std()
)


cv_comparison = pd.DataFrame({

    "Model": [

        "Text + Metadata",

        "Text + Metadata + Image"
    ],

    "Mean_CV_Accuracy": [

        baseline_cv_accuracy,

        multimodal_cv_accuracy
    ],

    "CV_Accuracy_Std": [

        baseline_cv_accuracy_std,

        multimodal_cv_accuracy_std
    ],

    "Mean_CV_Weighted_F1": [

        baseline_cv_f1,

        multimodal_cv_f1
    ],

    "CV_F1_Std": [

        baseline_cv_f1_std,

        multimodal_cv_f1_std
    ],

    "CV_Time_Seconds": [

        baseline_cv_time,

        multimodal_cv_time
    ]
})


print("\n" + "=" * 75)
print("CROSS-VALIDATION COMPARISON")
print("=" * 75)

display(
    cv_comparison.round(4)
)


cv_comparison_path = (
    RESULTS_DIR
    / "multimodal_cross_validation_comparison.csv"
)

cv_comparison.to_csv(
    cv_comparison_path,
    index=False
)


# ============================================================
# 14.21 CV IMPROVEMENT
# ============================================================

cv_accuracy_change = (
    multimodal_cv_accuracy
    - baseline_cv_accuracy
)

cv_f1_change = (
    multimodal_cv_f1
    - baseline_cv_f1
)


print("\n" + "=" * 75)
print("CROSS-VALIDATED IMAGE CONTRIBUTION")
print("=" * 75)

print(
    f"\nCV Accuracy change:"
    f" {cv_accuracy_change * 100:+.2f} percentage points"
)

print(
    f"CV Weighted F1 change:"
    f" {cv_f1_change * 100:+.2f} percentage points"
)


# ============================================================
# 14.22 SELECT FINAL MODEL
# ============================================================

print("\n" + "=" * 75)
print("FINAL MODEL SELECTION")
print("=" * 75)


# Primary criterion:
# Cross-validated weighted F1
#
# Secondary criterion:
# Cross-validated accuracy


if (

    multimodal_cv_f1 > baseline_cv_f1

    or (

        np.isclose(
            multimodal_cv_f1,
            baseline_cv_f1
        )

        and multimodal_cv_accuracy
        > baseline_cv_accuracy

    )
):

    selected_model_name = (
        "Multimodal HistGradientBoosting"
    )

    selected_pipeline = (
        multimodal_pipeline
    )

    selected_features = (
        multimodal_features
    )

    selected_accuracy = (
        multimodal_accuracy
    )

    selected_f1 = (
        multimodal_f1
    )

    selected_cv_accuracy = (
        multimodal_cv_accuracy
    )

    selected_cv_f1 = (
        multimodal_cv_f1
    )

    selection_reason = (
        "Image-derived features improved "
        "or maintained the primary "
        "cross-validated performance."
    )

else:

    selected_model_name = (
        "HistGradientBoosting Baseline"
    )

    selected_pipeline = (
        baseline_pipeline
    )

    selected_features = (
        baseline_features
    )

    selected_accuracy = (
        baseline_accuracy
    )

    selected_f1 = (
        baseline_f1
    )

    selected_cv_accuracy = (
        baseline_cv_accuracy
    )

    selected_cv_f1 = (
        baseline_cv_f1
    )

    selection_reason = (
        "The baseline model remained superior "
        "under cross-validation."
    )


print(
    "\nSelected model:",
    selected_model_name
)

print(
    "\nReason:",
    selection_reason
)

print(
    "\nHold-out Accuracy:",
    f"{selected_accuracy:.4%}"
)

print(
    "Hold-out Weighted F1:",
    f"{selected_f1:.4%}"
)

print(
    "Mean CV Accuracy:",
    f"{selected_cv_accuracy:.4%}"
)

print(
    "Mean CV Weighted F1:",
    f"{selected_cv_f1:.4%}"
)


# ============================================================
# 14.23 SAVE SELECTED MODEL
# ============================================================

FINAL_MODEL_PATH = (
    MODELS_DIR
    / "final_multimodal_or_baseline_model.joblib"
)


joblib.dump(
    selected_pipeline,
    FINAL_MODEL_PATH
)


print(
    "\nSelected model saved:"
)

print(
    FINAL_MODEL_PATH
)


# ============================================================
# 14.24 SAVE IMAGE-INTEGRATION METADATA
# ============================================================

metadata = {

    "experiment":
        "Multimodal Image Feature Integration",

    "dataset":
        "Synthetic V2",

    "dataset_rows":
        int(len(df)),

    "dataset_columns":
        int(len(df.columns)),

    "target":
        TARGET,

    "image_features_used":
        image_features_found,

    "image_feature_count":
        int(len(image_features_found)),

    "baseline_feature_count":
        int(len(baseline_features)),

    "multimodal_feature_count":
        int(len(multimodal_features)),

    "baseline_holdout_accuracy":
        float(baseline_accuracy),

    "multimodal_holdout_accuracy":
        float(multimodal_accuracy),

    "baseline_holdout_weighted_f1":
        float(baseline_f1),

    "multimodal_holdout_weighted_f1":
        float(multimodal_f1),

    "baseline_cv_accuracy":
        float(baseline_cv_accuracy),

    "multimodal_cv_accuracy":
        float(multimodal_cv_accuracy),

    "baseline_cv_weighted_f1":
        float(baseline_cv_f1),

    "multimodal_cv_weighted_f1":
        float(multimodal_cv_f1),

    "cv_accuracy_change":
        float(cv_accuracy_change),

    "cv_weighted_f1_change":
        float(cv_f1_change),

    "selected_model":
        selected_model_name,

    "selected_holdout_accuracy":
        float(selected_accuracy),

    "selected_holdout_weighted_f1":
        float(selected_f1),

    "selected_cv_accuracy":
        float(selected_cv_accuracy),

    "selected_cv_weighted_f1":
        float(selected_cv_f1),

    "leakage_check":
        "PASSED",

    "random_state":
        42
}


METADATA_PATH = (
    RESULTS_DIR
    / "multimodal_image_integration_metadata.json"
)


with open(
    METADATA_PATH,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        metadata,
        f,
        indent=4
    )


# ============================================================
# 14.25 FINAL ACADEMIC SUMMARY
# ============================================================

print("\n")
print("=" * 75)
print("MULTIMODAL IMAGE INTEGRATION COMPLETED")
print("=" * 75)

print(
    "\nIMAGE FEATURES USED:"
)

for feature in image_features_found:

    print(
        "  -",
        feature
    )


print(
    "\nBASELINE MODEL"
)

print(
    f"Hold-out Accuracy : "
    f"{baseline_accuracy:.4%}"
)

print(
    f"Hold-out F1       : "
    f"{baseline_f1:.4%}"
)

print(
    f"Mean CV Accuracy  : "
    f"{baseline_cv_accuracy:.4%}"
)

print(
    f"Mean CV F1        : "
    f"{baseline_cv_f1:.4%}"
)


print(
    "\nMULTIMODAL MODEL"
)

print(
    f"Hold-out Accuracy : "
    f"{multimodal_accuracy:.4%}"
)

print(
    f"Hold-out F1       : "
    f"{multimodal_f1:.4%}"
)

print(
    f"Mean CV Accuracy  : "
    f"{multimodal_cv_accuracy:.4%}"
)

print(
    f"Mean CV F1        : "
    f"{multimodal_cv_f1:.4%}"
)


print(
    "\nIMAGE CONTRIBUTION"
)

print(
    f"Hold-out Accuracy Change : "
    f"{accuracy_change * 100:+.2f} pp"
)

print(
    f"CV Accuracy Change       : "
    f"{cv_accuracy_change * 100:+.2f} pp"
)

print(
    f"CV Weighted F1 Change    : "
    f"{cv_f1_change * 100:+.2f} pp"
)


print(
    "\nFINAL SELECTED MODEL:"
)

print(
    selected_model_name
)


print(
    "\nFinal hold-out accuracy:"
)

print(
    f"{selected_accuracy:.4%}"
)


print(
    "\nFinal mean CV accuracy:"
)

print(
    f"{selected_cv_accuracy:.4%}"
)


print(
    "\nFinal mean CV Weighted F1:"
)

print(
    f"{selected_cv_f1:.4%}"
)


print(
    "\nSelection reason:"
)

print(
    selection_reason
)


print(
    "\nSaved model:"
)

print(
    FINAL_MODEL_PATH
)


print(
    "\nSaved metadata:"
)

print(
    METADATA_PATH
)


print("\n" + "=" * 75)
print("READY FOR REAL-DATA MULTIMODAL EVALUATION")
print("=" * 75)
```
